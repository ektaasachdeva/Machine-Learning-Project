## **2. Dataset Ingestion (UCI Adult Dataset)**
We implement a programmatic multi-tier ingestion routine using ucimlrepo with direct web fallbacks.

```python
def load_adult_dataset():
    column_names = [
        'age', 'workclass', 'fnlwgt', 'education', 'education-num',
        'marital-status', 'occupation', 'relationship', 'race', 'sex',
        'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income'
    ]
    try:
        from ucimlrepo import fetch_ucirepo
        print('[INFO] Querying UCI Repository via API (ID=2)... ')
        dataset = fetch_ucirepo(id=2)
        X = dataset.data.features
        y = dataset.data.targets
        df = pd.concat([X, y], axis=1)
        print(f'[SUCCESS] Loaded via ucimlrepo API. Dimensions: {df.shape}')
        return df
    except Exception as e:
        print(f'[NOTE] ucimlrepo query skipped ({e}). Trying direct URL...')

    try:
        train_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data'
        test_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.test'
        train_df = pd.read_csv(train_url, header=None, names=column_names, na_values=' ?', skipinitialspace=True)
        test_df = pd.read_csv(test_url, header=None, names=column_names, na_values=' ?', skipinitialspace=True, skiprows=1)
        df = pd.concat([train_df, test_df], ignore_index=True)
        print(f'[SUCCESS] Loaded via direct UCI repository URLs. Dimensions: {df.shape}')
        return df
    except Exception as e:
        print(f'[NOTE] Web download failed ({e}). Checking local file...')

    if os.path.exists('adult.csv'):
        df = pd.read_csv('adult.csv', skipinitialspace=True, na_values='?')
        return df

    raise RuntimeError('Dataset could not be loaded. Please upload adult.csv.')

df_raw = load_adult_dataset()
```

## **3. Data Inspection, Cleaning & Preprocessing**
We inspect raw data, clean strings, standardize the target variable, eliminate duplicate rows, and handle missing values.

```python
print('=== FIRST 5 RECORDS ===')
display(df_raw.head())

print(f'\nDataset Dimensions: {df_raw.shape[0]} rows, {df_raw.shape[1]} columns')
print('\n=== DATA TYPES ===')
print(df_raw.dtypes)

# Clean string whitespaces
df_clean = df_raw.copy()
for col in df_clean.select_dtypes(include=['object']).columns:
    df_clean[col] = df_clean[col].astype(str).str.strip()

# Standardize target variable (strip trailing period from adult.test)
df_clean['income'] = df_clean['income'].str.rstrip('.')
df_clean['income_binary'] = df_clean['income'].apply(lambda x: 1 if x == '>50K' else 0)

# Handle '?' missing values
df_clean.replace('?', np.nan, inplace=True)
missing = df_clean.isnull().sum()
print('\n=== MISSING VALUES PER COLUMN ===')
print(missing[missing > 0])

# Duplicate removal
dups = df_clean.duplicated().sum()
print(f'\nDuplicate records found: {dups}')
if dups > 0:
    df_clean.drop_duplicates(inplace=True)
    print(f'Shape after duplicate removal: {df_clean.shape}')

# Categorical missing value imputation with Unknown category
for col in ['workclass', 'occupation', 'native-country']:
    if col in df_clean.columns:
        df_clean[col].fillna('Unknown', inplace=True)

# Sensitive attribute distribution
print('\n=== SENSITIVE ATTRIBUTE (sex) DISTRIBUTION ===')
print(df_clean['sex'].value_counts(normalize=True).apply(lambda p: f'{p*100:.2f}%'))
```

## **5. Preprocessing Pipelines & Leakage-Free Partitions**
We construct an 80-20 stratified split based on the joint distribution of target class and sensitive attribute. Preprocessors are fitted strictly on the training partition.

```python
X = df_clean.drop(columns=['income', 'income_binary', 'fnlwgt', 'education'])
y = df_clean['income_binary'].values
sensitive_attr = df_clean['sex'].values

num_features = ['age', 'education-num', 'capital-gain', 'capital-loss', 'hours-per-week']
cat_features = [c for c in X.columns if c not in num_features]

joint_strata = df_clean['sex'] + '_' + df_clean['income_binary'].astype(str)
X_train, X_test, y_train, y_test, s_train, s_test = train_test_split(
    X, y, sensitive_attr, test_size=0.20, random_state=42, stratify=joint_strata
)

num_trans = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
cat_trans = Pipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer([
    ('num', num_trans, num_features),
    ('cat', cat_trans, cat_features)
])

cat_no_sex = [c for c in cat_features if c != 'sex']
preprocessor_no_sex = ColumnTransformer([
    ('num', num_trans, num_features),
    ('cat', cat_trans, cat_no_sex)
])
```

## **6. Machine Learning Classification & Benchmark**
We train and benchmark four classifiers across 5-Fold Stratified Cross-Validation: Logistic Regression, Decision Tree Classifier, Gaussian Naive Bayes, and K-Nearest Neighbours.

```python
models = {
    'Logistic Regression': LogisticRegression(C=1.0, max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(max_depth=7, min_samples_leaf=20, random_state=42),
    'Gaussian Naive Bayes': GaussianNB(),
    'K-Nearest Neighbours': KNeighborsClassifier(n_neighbors=15, n_jobs=-1)
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
benchmark_list = []
trained_pipelines = {}
roc_dict = {}

for name, clf in models.items():
    pipeline = Pipeline([('preprocessor', preprocessor), ('clf', clf)])
    cv_res = cross_validate(pipeline, X_train, y_train, cv=cv, scoring=['accuracy', 'f1', 'roc_auc'], n_jobs=-1)
    pipeline.fit(X_train, y_train)
    trained_pipelines[name] = pipeline
    
    y_pred = pipeline.predict(X_test)
    y_proba = pipeline.predict_proba(X_test)[:, 1] if hasattr(pipeline, 'predict_proba') else y_pred
    
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    auc = roc_auc_score(y_test, y_proba)
    
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    roc_dict[name] = (fpr, tpr, auc)
    
    benchmark_list.append({
        'Model': name,
        'Accuracy': acc,
        'Precision': prec,
        'Recall': rec,
        'F1-Score': f1,
        'ROC-AUC': auc,
        'CV Accuracy': f"{cv_res['test_accuracy'].mean():.4f} Â± {cv_res['test_accuracy'].std():.4f}",
        'CV F1': f"{cv_res['test_f1'].mean():.4f} Â± {cv_res['test_f1'].std():.4f}"
    })

benchmark_df = pd.DataFrame(benchmark_list)
display(benchmark_df)
```

```python
# Figure 8: ROC Curves
plt.figure(figsize=(7, 5))
for name, (fpr, tpr, auc) in roc_dict.items():
    plt.plot(fpr, tpr, lw=2, label=f'{name} (AUC = {auc:.3f})')
plt.plot([0, 1], [0, 1], 'k--', lw=1.2, label='Random Chance (0.500)')
plt.title('Figure 8: ROC Curves on Test Set', fontweight='bold')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()

# Figure 9: Confusion Matrices
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.ravel()
for idx, (name, pipe) in enumerate(trained_pipelines.items()):
    y_p = pipe.predict(X_test)
    cm = confusion_matrix(y_test, y_p)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[idx], cbar=False,
                xticklabels=['<=50K', '>50K'], yticklabels=['<=50K', '>50K'])
    axes[idx].set_title(f'{name}\nAcc: {accuracy_score(y_test, y_p):.3f} | F1: {f1_score(y_test, y_p):.3f}', fontweight='bold')
    axes[idx].set_xlabel('Predicted')
    axes[idx].set_ylabel('True')
plt.suptitle('Figure 9: Confusion Matrices', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## **7. Gender Fairness Audit**
We calculate Selection Rates, Demographic Parity Difference, Disparate Impact Ratio, Equal Opportunity Difference, and Error Rate Disparities across all four baseline classifiers.

```python
def audit_fairness(y_true, y_pred, sensitive_features, priv='Male', unpriv='Female'):
    mask_m = (sensitive_features == priv)
    mask_f = (sensitive_features == unpriv)
    
    sr_m = np.mean(y_pred[mask_m])
    sr_f = np.mean(y_pred[mask_f])
    dpd = sr_m - sr_f
    di = (sr_f / sr_m) if sr_m > 0 else np.nan
    
    cm_m = confusion_matrix(y_true[mask_m], y_pred[mask_m], labels=[0, 1])
    tn_m, fp_m, fn_m, tp_m = cm_m.ravel()
    tpr_m = tp_m / (tp_m + fn_m) if (tp_m + fn_m) > 0 else 0
    fpr_m = fp_m / (fp_m + tn_m) if (fp_m + tn_m) > 0 else 0
    fnr_m = 1.0 - tpr_m
    
    cm_f = confusion_matrix(y_true[mask_f], y_pred[mask_f], labels=[0, 1])
    tn_f, fp_f, fn_f, tp_f = cm_f.ravel()
    tpr_f = tp_f / (tp_f + fn_f) if (tp_f + fn_f) > 0 else 0
    fpr_f = fp_f / (fp_f + tn_f) if (fp_f + tn_f) > 0 else 0
    fnr_f = 1.0 - tpr_f
    
    eopp = tpr_m - tpr_f
    eodds = max(abs(fpr_m - fpr_f), abs(tpr_m - tpr_f))
    
    return {
        'Selection Rate (Male)': sr_m,
        'Selection Rate (Female)': sr_f,
        'Disparate Impact Ratio': di,
        'Demographic Parity Diff': dpd,
        'TPR / Recall (Male)': tpr_m,
        'TPR / Recall (Female)': tpr_f,
        'Equal Opportunity Diff': eopp,
        'FNR (Male)': fnr_m,
        'FNR (Female)': fnr_f,
        'Equalized Odds Diff': eodds
    }

audit_list = []
for name, pipe in trained_pipelines.items():
    y_p = pipe.predict(X_test)
    res = audit_fairness(y_test, y_p, s_test)
    res['Model'] = name
    audit_list.append(res)

audit_df = pd.DataFrame(audit_list)
cols_to_show = ['Model', 'Selection Rate (Male)', 'Selection Rate (Female)', 'Disparate Impact Ratio', 'Demographic Parity Diff', 'Equal Opportunity Diff', 'FNR (Male)', 'FNR (Female)', 'Equalized Odds Diff']
display(audit_df[cols_to_show])
```

## **8. Bias Mitigation Interventions & Pareto Trade-Offs**
We compare: Baseline Model vs. Fairness Through Unawareness vs. Kamiran & Calders Pre-processing Re-weighting vs. Post-processing Threshold Optimization.

```python
# 1. Baseline Model
pipe_base = trained_pipelines['Logistic Regression']
y_p_base = pipe_base.predict(X_test)
y_pr_base = pipe_base.predict_proba(X_test)[:, 1]
r_base = audit_fairness(y_test, y_p_base, s_test)
r_base['Intervention'] = '1. Baseline (Unmitigated)'
r_base['Accuracy'] = accuracy_score(y_test, y_p_base)
r_base['F1-Score'] = f1_score(y_test, y_p_base)
r_base['ROC-AUC'] = roc_auc_score(y_test, y_pr_base)

# 2. Fairness Through Unawareness (dropping sex)
X_tr_nosex = X_train.drop(columns=['sex'])
X_te_nosex = X_test.drop(columns=['sex'])
pipe_unaware = Pipeline([('preprocessor', preprocessor_no_sex), ('clf', LogisticRegression(C=1.0, max_iter=1000, random_state=42))])
pipe_unaware.fit(X_tr_nosex, y_train)
y_p_unaware = pipe_unaware.predict(X_te_nosex)
y_pr_unaware = pipe_unaware.predict_proba(X_te_nosex)[:, 1]
r_unaware = audit_fairness(y_test, y_p_unaware, s_test)
r_unaware['Intervention'] = '2. Unawareness (Drop Sex)'
r_unaware['Accuracy'] = accuracy_score(y_test, y_p_unaware)
r_unaware['F1-Score'] = f1_score(y_test, y_p_unaware)
r_unaware['ROC-AUC'] = roc_auc_score(y_test, y_pr_unaware)

# 3. Kamiran & Calders Pre-processing Reweighing
def get_weights(y, s, priv='Male', unpriv='Female'):
    w = np.ones(len(y))
    p_sm, p_sf = np.mean(s == priv), np.mean(s == unpriv)
    p_y1, p_y0 = np.mean(y == 1), np.mean(y == 0)
    for s_v, p_s in [(priv, p_sm), (unpriv, p_sf)]:
        for y_v, p_y in [(1, p_y1), (0, p_y0)]:
            mask = (s == s_v) & (y == y_v)
            pj = np.mean(mask)
            if pj > 0:
                w[mask] = (p_s * p_y) / pj
    return w

weights_train = get_weights(y_train, s_train)
pipe_reweighted = Pipeline([('preprocessor', preprocessor), ('clf', LogisticRegression(C=1.0, max_iter=1000, random_state=42))])
pipe_reweighted.fit(X_train, y_train, clf__sample_weight=weights_train)
y_p_rw = pipe_reweighted.predict(X_test)
y_pr_rw = pipe_reweighted.predict_proba(X_test)[:, 1]
r_rw = audit_fairness(y_test, y_p_rw, s_test)
r_rw['Intervention'] = '3. Sample Re-weighting'
r_rw['Accuracy'] = accuracy_score(y_test, y_p_rw)
r_rw['F1-Score'] = f1_score(y_test, y_p_rw)
r_rw['ROC-AUC'] = roc_auc_score(y_test, y_pr_rw)

# 4. Post-processing Thresholding
y_p_post = np.zeros_like(y_test)
y_p_post[(s_test == 'Male') & (y_pr_base >= 0.58)] = 1
y_p_post[(s_test == 'Female') & (y_pr_base >= 0.35)] = 1
r_post = audit_fairness(y_test, y_p_post, s_test)
r_post['Intervention'] = '4. Group Thresholding'
r_post['Accuracy'] = accuracy_score(y_test, y_p_post)
r_post['F1-Score'] = f1_score(y_test, y_p_post)
r_post['ROC-AUC'] = roc_auc_score(y_test, y_pr_base)

mit_df = pd.DataFrame([r_base, r_unaware, r_rw, r_post])
table_c = ['Intervention', 'Accuracy', 'F1-Score', 'ROC-AUC', 'Disparate Impact Ratio', 'Demographic Parity Diff', 'Equal Opportunity Diff', 'Equalized Odds Diff']
display(mit_df[table_c])
```

```python
# Figure 11: Fairness-Utility Trade-off Visualisations
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Panel A: F1-Score vs Disparate Impact Ratio
for _, row in mit_df.iterrows():
    axes[0].scatter(row['F1-Score'], row['Disparate Impact Ratio'], s=160, zorder=5)
    axes[0].annotate(row['Intervention'], (row['F1-Score'] + 0.002, row['Disparate Impact Ratio'] - 0.02), fontweight='bold', fontsize=9)
axes[0].axhline(0.80, color='red', linestyle='--', label='EEOC 80% Rule Threshold')
axes[0].set_title('Figure 11A: Utility (F1) vs Disparate Impact Ratio', fontweight='bold')
axes[0].set_xlabel('F1-Score (Predictive Utility)')
axes[0].set_ylabel('Disparate Impact Ratio (Higher is Fairer)')
axes[0].legend(loc='lower left')

# Panel B: Accuracy vs Equal Opportunity Difference
for _, row in mit_df.iterrows():
    axes[1].scatter(row['Accuracy'], row['Equal Opportunity Diff'], s=160, zorder=5)
    axes[1].annotate(row['Intervention'], (row['Accuracy'] + 0.001, row['Equal Opportunity Diff'] + 0.007), fontweight='bold', fontsize=9)
axes[1].axhline(0.0, color='green', linestyle='--', label='Ideal Parity (0.0)')
axes[1].set_title('Figure 11B: Utility (Accuracy) vs Equal Opportunity Gap', fontweight='bold')
axes[1].set_xlabel('Accuracy')
axes[1].set_ylabel('Equal Opportunity Gap (Closer to 0 is Fairer)')
axes[1].legend(loc='upper right')

plt.tight_layout()
plt.show()
```
